#  Coffee Store Analysis Dashboard (Databricks SQL)

This project includes a Databricks SQL dashboard called **Coffee Store Analysis** built on top of the **Gold layer star schema** and **Gold aggregated materialized views**.

The dashboard provides an executive summary of store sales performance, transaction volume, customer trends, and product insights.

---

##  Dashboard Objective

The dashboard is designed for **business stakeholders (Finance / Leadership)** to monitor:

- Overall revenue performance
- Monthly sales and transaction trends
- Average order value behavior
- Payment method preferences
- Top performing stores
- Best selling menu items

---

##  Data Sources (Gold Layer)

The dashboard and analytics layer uses the following Gold tables and materialized views.

### Fact Tables
- `coffee.gold.fact_transactions`
- `coffee.gold.fact_transaction_items`

### Dimensions
- `coffee.gold.dim_stores`
- `coffee.gold.dim_menu_items`
- `coffee.gold.dim_users`

---

##  Materialized Views (Aggregations)

### Views used in the dashboard
- `coffee.gold.agg_monthly_sales`
- `coffee.gold.agg_payment_method_split`
- `coffee.gold.agg_store_performance`
- `coffee.gold.agg_top_3_menu_items`

### Additional views created for advanced analytics
- `coffee.gold.agg_voucher_usage`
- `coffee.gold.agg_customer_clv`
- `coffee.gold.agg_top_10_customers_by_spend`
- `coffee.gold.agg_customer_churn_6m`
- `coffee.gold.agg_new_customer_churn_6m`

These additional views represent an analytics layer on top of the Gold facts, and can be reused for dashboards, reporting, or ad-hoc analysis.

## 🔐 Security Note (Unity Catalog)

Row Level Security (RLS) and Column Level Security (CLS) are implemented in this project using Unity Catalog.

- Store managers are restricted to their assigned store using RLS.
- Finance users have full access.
- PII columns in `dim_users` are protected using CLS masking.

This dashboard is primarily designed for **Finance users**, and is expected to be shared using Finance-level permissions.

If shared with store manager users, dashboard results will depend on the dashboard publish option:

- **Shared data permission** → dashboard runs using owner permissions
- **Individual data permission** → dashboard runs using viewer permissions (RLS applies)
