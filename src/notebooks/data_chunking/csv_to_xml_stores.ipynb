### XML Data Ingestion – Stores Dataset

Purpose
This notebook converts the `stores` dataset from CSV format to XML format
as part of demonstrating support for multiple data formats in the ingestion layer.
The conversion is done using PySpark read/write APIs and stores the output in a Unity Catalog Volume.

Storage
- Input: CSV files stored in Unity Catalog Volume
- Output: XML files written back to the same volume under a separate folder

 Key Notes
- Databricks UI does not support previewing XML files stored in Volumes.
- This limitation does not affect Spark's ability to read, validate, or process XML data.







```python
# Read the raw stores CSV dataset from UC Volume.
# This is a pre-processing step to convert the dataset into XML
# for demonstrating XML ingestion in the Bronze layer.

df = (
    spark.read
    .option("header", True)
    .csv("/Volumes/workspace/default/coffee_raw_volume/stores/")
)

```

```python

# Write the dataset as XML using Spark XML writer.
# rootTag defines the top-level XML container (<stores>),
# rowTag defines one row element (<store>) per record.
# Overwrite mode makes the conversion reproducible on reruns.
(
    df.write
    .mode("overwrite")
    .format("xml")
    .option("rootTag", "stores")
    .option("rowTag", "store")
    .save("/Volumes/workspace/default/coffee_raw_volume/stores_xml/")
)
```
