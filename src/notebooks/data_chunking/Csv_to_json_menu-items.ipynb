```python
# Read menu_items raw CSV from UC Volume.
# This is a pre-processing step to convert the dataset into JSON
# so that Auto Loader ingestion can be demonstrated in the Bronze layer.
df = (
    spark.read
    .option("header", True)
    .csv("/Volumes/workspace/default/coffee_raw_volume/menu_items/")
)
```

```python
# Write menu_items into JSON format (overwrite mode).
# This ensures the output folder remains clean and reproducible on reruns.

(
    df
    .write
    .mode("overwrite")
    .json("/Volumes/workspace/default/coffee_raw_volume/menu_items_json/")
)
```

```python
# Validate that JSON files were created successfully in the target folder.
dbutils.fs.ls("/Volumes/workspace/default/coffee_raw_volume/menu_items_json/")
```
