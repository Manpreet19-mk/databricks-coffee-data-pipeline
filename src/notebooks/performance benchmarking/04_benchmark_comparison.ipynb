# Gold Layer Performance Benchmark - Final Comparison

This notebook summarizes the benchmark results for 3 optimization strategies on Gold layer fact tables.

### Strategies Compared
1. **Baseline** (No optimization)
2. **Liquid Clustering**
3. **Partitioning + Z-ORDER**

### Benchmark Queries Used
- Q1: Monthly Sales Trend  
- Q2: Store Performance  
- Q3: Top 10 Customers by Spend  
- Q4: Top 10 Menu Items by Revenue  
- Q5: Join Drilldown (Store + Item Revenue)

> Note: All runtimes recorded are from **Run 2** to avoid warm-up overhead.


%md
## Benchmark Results (Baseline vs Liquid Clustering vs Partition + ZORDER)

| Query ID | Query Name | Baseline Runtime (Run 2) | Liquid Clustering Runtime (Run 2) | Partition + ZORDER Runtime (Run 2) |
|---------|------------|---------------------------|-----------------------------------|------------------------------------|
| Q1 | Monthly Sales Trend | 2.23 | 1.37 | 1.54 |

%md
## Observations

### Key Findings
- **Liquid Clustering** gave the best improvement for **join-heavy queries (Q5)**.
- **Partition + ZORDER** performed slightly better for **customer ranking query (Q3)**.
- Overall, both approaches improved performance compared to baseline.

### Conclusion
For this dataset, **Liquid Clustering** is the best overall strategy because it improves performance consistently without needing heavy partition management.
