## GOLD LAYER - DIMENSIONS (SCD TYPE 2)

### Purpose
- Create Gold dimension tables from Gold staging tables  
- Maintain full history of changes using **SCD Type 2**  
- Use **silver_updated_at** as the sequencing column for versioning  

### Output Tables
- `coffee.gold.dim_users`
- `coffee.gold.dim_stores`
- `coffee.gold.dim_menu_items`
- `coffee.gold.dim_vouchers`
- `coffee.gold.dim_payment_methods`

### Notes
- DLT automatically adds SCD2 metadata columns:
  - `__START_AT`
  - `__END_AT`
  - `__CURRENT`


```sql
-- DIM: USERS (SCD2)
-- Key: user_id

CREATE OR REFRESH STREAMING TABLE coffee.gold.dim_users
COMMENT "Gold dimension: users (SCD Type 2) built from stg_users";

APPLY CHANGES INTO coffee.gold.dim_users
FROM STREAM(coffee.gold.stg_users)
KEYS (user_id)
SEQUENCE BY silver_updated_at
STORED AS SCD TYPE 2;
```
