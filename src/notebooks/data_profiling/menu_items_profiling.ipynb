## # Data Profiling – Menu Items

This notebook profiles the menu_items dataset to understand its schema,
cardinality, data quality, and overall structure in order to inform
downstream design decisions.


```python
%python
# understanding the schema and data types
df = (
    spark.read
    .option("header", True)
    .csv("/Volumes/workspace/default/coffee_raw_volume/menu_items/")
)

df.printSchema()
df.show()
```

```python
# Basic Counts
from pyspark.sql.functions import count , countDistinct

df.select(
    count("*").alias("total_rows"),
    countDistinct("item_id").alias("distinct_item_ids"),
    countDistinct("item_name").alias("distinct_item_names")
).show()
```

```python
# Category Analysis
df.groupBy("category").count().show()
```

```python
# Seasonal Flag check
df.groupBy("is_seasonal").count().show()
 
```

```python
# Null analysis
df.selectExpr(
    "sum(case when item_id is null then 1 else 0 end) as null_item_id",
    "sum(case when item_name is null then 1 else 0 end) as null_item_name",
    "sum(case when category is null then 1 else 0 end) as null_category",
    "sum(case when price is null then 1 else 0 end) as null_price"
).show()
```

### Key observations
- menu_items contains 8 records
- itme_id and item_name are unique
- No null columns found 
- Category has coffee and non-coffee itmes
- is_seasonal column exists but contains all false values
- Columns available_from and available_to exist in the schema but unused in the current dataset
