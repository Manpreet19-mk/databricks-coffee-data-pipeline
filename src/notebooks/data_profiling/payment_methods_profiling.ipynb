###  Data Profiling – Payment Methods

This notebook profiles the payment_methods dataset to understand its schema,
cardinality, and data quality.


```python
# reading data from volume
df = (
    spark.read
    .option("header", True)
    .csv("/Volumes/workspace/default/coffee_raw_volume/payment_methods/")
)

df.printSchema()
df.show()
```

```python
# Row count & Uniqueness
from pyspark.sql.functions import count, countDistinct

df.select(
    count("*").alias("total_rows"),
    countDistinct("method_id").alias("distinct_method_ids"),
    countDistinct("method_name").alias("distinct_method_names")
).show()
```

```python
# Null checks
df.selectExpr(
    "sum(case when method_id is null then 1 else 0 end) as null_method_id",
    "sum(case when method_name is null then 1 else 0 end) as null_method_name",
    "sum(case when category is null then 1 else 0 end) as null_category"
).show()
```

### Observations
- Payment methods contains 5 records
- method_id and method_name are unique
- no null values found in any column
- dataset is small and static in nature
