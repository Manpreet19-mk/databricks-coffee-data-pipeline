### # Data Profiling – Users

This notebook profiles the users dataset to understand user attributes,
registration patterns, and data quality.


```python
# reading all user files to a dataframe
df = (
    spark.read
    .option("header", True)
    .csv("/Volumes/workspace/default/coffee_raw_volume/users/")
)

df.printSchema()
df.show(5)
```

```python
# row count & uniqueness
from pyspark.sql.functions import count, countDistinct

df.select(
    count("*").alias("total_users"),
    countDistinct("user_id").alias("distinct_user_ids")
).show()
```

```python
# gender distribution
df.groupBy("gender").count().show()
```

```python
# registration date range
from pyspark.sql.functions import min, max

df.select(
    min("registered_at").alias("min_registered_at"),
    max("registered_at").alias("max_registered_at")
).show()
```

```python
# Null checks
df.selectExpr(
    "sum(case when user_id is null then 1 else 0 end) as null_user_id",
    "sum(case when gender is null then 1 else 0 end) as null_gender",
    "sum(case when birthdate is null then 1 else 0 end) as null_birthdate",
    "sum(case when registered_at is null then 1 else 0 end) as null_registered_at"
).show()
```

### Key Observations

- The users dataset contains approximately 2.2 million records with unique user_id values.
- user_id is globally unique across the dataset with no duplicate records observed.
- Gender distribution is nearly evenly split between male and female users.
- User registrations span from July 2023 to June 2025, indicating continuous growth over time.
- The users dataset represents registered customers, while transactions may occur without user association, indicating guest purchases.
