###  Data Profiling – Stores

This notebook profiles the stores dataset to understand its schema,
cardinality, and data quality.



```python
# reading data from volume
df = (
    spark.read
    .option("header", True)
    .csv("/Volumes/workspace/default/coffee_raw_volume/stores/")
)

df.printSchema()
df.show()
```

```python
# Row count and uniqueness
from pyspark.sql.functions import count, countDistinct

df.select(
    count("*").alias("total_rows"),
    countDistinct("store_id").alias("distinct_store_ids"),
    countDistinct("store_name").alias("distinct_store_names")
).show()
```

```python
# Null checks
df.selectExpr(
    "sum(case when store_id is null then 1 else 0 end) as null_store_id",
    "sum(case when store_name is null then 1 else 0 end) as null_store_name",
    "sum(case when city is null then 1 else 0 end) as null_city",
    "sum(case when postal_code is null then 1 else 0 end) as null_postal_code"
).show()
```

### Key Observations
- Stores contains 10 recods
- each store has a defined city and postal code
- No null values found in key columns
- dataset is small and static in nature
