####  Bronze vs Silver Testing – menu_items Table

This document describes the data quality, integrity, and reconciliation tests performed for the `menu_items` table during ingestion from Bronze to Silver.





####  Tables Under Test

- Bronze: `coffee.bronze.menu_items`
- Silver: `coffee.silver.menu_items`


```sql
-- Test 1: Compare row counts between Bronze and Silver
-- Silver count should be <= Bronze count due to filtering and deduplication

SELECT 'bronze' AS layer, COUNT(*) AS record_count
FROM coffee.bronze.menu_items

UNION ALL

SELECT 'silver' AS layer, COUNT(*) AS record_count
FROM coffee.silver.menu_items;
```

```sql
-- Test 2: Ensure mandatory columns are NOT NULL in Silver
-- item_id, item_name, category, and price must always be present

SELECT COUNT(*) AS invalid_silver_records
FROM coffee.silver.menu_items
WHERE
  item_id IS NULL
  OR item_name IS NULL
  OR category IS NULL
  OR price IS NULL;
```

```sql
-- Test 3: Validate menu item prices
-- Prices should be greater than zero

SELECT COUNT(*) AS invalid_prices
FROM coffee.silver.menu_items
WHERE price <= 0;
```

```sql
-- Test 4: All valid Bronze menu items should be present in Silver
-- Identifies valid records accidentally dropped

SELECT item_id
FROM coffee.bronze.menu_items
WHERE
  item_id IS NOT NULL
  AND item_name IS NOT NULL
  AND category IS NOT NULL
  AND price IS NOT NULL

EXCEPT

SELECT item_id
FROM coffee.silver.menu_items;
```
