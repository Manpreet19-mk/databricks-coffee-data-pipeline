####Unit Testing – Bronze stores
This notebook validates the Bronze ingestion of stores data by comparing the ingested Bronze table with the raw source files stored in Databricks Volumes.

The tests focus on:

- Row count consistency
- Null checks
- Duplicate detection

```sql
-- Row count comparison between Bronze table and original CSV source
SELECT
  'bronze_table' AS source_type,
  COUNT(*) AS row_count
FROM coffee.bronze.stores

UNION ALL

SELECT
  'raw_volume' AS source_type,
  COUNT(*) AS row_count
FROM (
  SELECT *
  FROM read_files(
    '/Volumes/workspace/default/coffee_raw_volume/stores',
    format => 'csv',
    header => true
  )
);
```

```sql
SELECT
  COUNT(*) AS null_store_id_count
FROM coffee.bronze.stores
WHERE store_id IS NULL;
```

```sql
--duplicate check on store_id
SELECT
 store_id,
  COUNT(*) AS cnt
FROM coffee.bronze.stores
GROUP BY store_id
HAVING COUNT(*) > 1;
```
