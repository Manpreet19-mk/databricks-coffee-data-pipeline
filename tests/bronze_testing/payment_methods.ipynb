####Unit Testing – Bronzepayment methods
This notebook validates the Bronze ingestion of payment methods data by comparing the ingested Bronze table with the raw source files stored in Databricks Volumes.

The tests focus on:

- Row count consistency
- Null checks
- Duplicate detection

```sql
-- Row count comparison between Bronze table and original CSV source
SELECT
  'bronze_table' AS source_type,
  COUNT(*) AS row_count
FROM coffee.bronze.payment_methods

UNION ALL

SELECT
  'raw_csv_volume' AS source_type,
  COUNT(*) AS row_count
FROM (
  SELECT *
  FROM read_files(
    '/Volumes/workspace/default/coffee_raw_volume/payment_methods',
    format => 'csv',
    header => true
  )
);
```

```sql
-- null value check
SELECT
  SUM(CASE WHEN method_id IS NULL THEN 1 ELSE 0 END) AS null_method_id_count,
  SUM(CASE WHEN method_name IS NULL THEN 1 ELSE 0 END) AS null_method_name_count
FROM coffee.bronze.payment_methods;
```

```sql
-- duplicate check
SELECT
 method_id,
 method_name,
  COUNT(*) AS cnt
FROM coffee.bronze.payment_methods
GROUP BY 1,2
HAVING COUNT(*) > 1;
```
