#### Unit Testing – Bronze Transaction_items
This notebook validates the Bronze ingestion of the Transaction_items dataset.
The source data consists of multiple CSV files (one per month), which are
ingested into a single Bronze table.

The objective of these tests is to ensure that:
- All records from all source files are ingested
- No records are lost or duplicated
- Key identifiers are present
- Ingestion completeness is verified


```sql
-- Row count comparison between raw volume and bronze table
SELECT
  'bronze_table' AS source_type,
  COUNT(*) AS row_count
FROM coffee.bronze.transaction_items

UNION ALL

SELECT
  'raw_csv_volume' AS source_type,
  COUNT(*) AS row_count
FROM (
  SELECT *
  FROM read_files(
    '/Volumes/workspace/default/coffee_raw_volume/transaction_items',
    format => 'csv',
    header => true
  )
);
```

```sql
-- comparison of sum of quantity, unit price and subtotal
SELECT
  'bronze_table' AS source_type,
  SUM(CAST(quantity AS DOUBLE))    AS total_quantity,
  SUM(CAST(unit_price AS DOUBLE))  AS total_unit_price,
  SUM(CAST(subtotal AS DOUBLE))    AS total_subtotal
FROM coffee.bronze.transaction_items

UNION ALL

SELECT
  'raw_csv_volume' AS source_type,
  SUM(CAST(quantity AS DOUBLE))    AS total_quantity,
  SUM(CAST(unit_price AS DOUBLE))  AS total_unit_price,
  SUM(CAST(subtotal AS DOUBLE))    AS total_subtotal
FROM (
  SELECT *
  FROM read_files(
    '/Volumes/workspace/default/coffee_raw_volume/transaction_items',
    format => 'csv',
    header => true
  )
);
```

```sql
-- null check on transaction_id
SELECT
  COUNT(*) AS null_transaction_id_count
FROM coffee.bronze.transaction_items
WHERE transaction_id IS NULL;
```

```sql
-- total file sources check
SELECT
  COUNT(DISTINCT source_file) AS distinct_source_files
FROM coffee.bronze.transaction_items;
```
