#### Unit Testing – Bronze Users
This notebook validates the Bronze ingestion of the users dataset.
The source data consists of multiple CSV files (one per month), which are
ingested into a single Bronze table.

The objective of these tests is to ensure that:
- All records from all source files are ingested
- No records are lost or duplicated
- Key identifiers are present
- Ingestion completeness is verified


```sql
-- Row count comparison between raw volume and bronze table
SELECT
  'bronze_table' AS source_type,
  COUNT(*) AS row_count
FROM coffee.bronze.users

UNION ALL

SELECT
  'raw_csv_volume' AS source_type,
  COUNT(*) AS row_count
FROM (
  SELECT *
  FROM read_files(
    '/Volumes/workspace/default/coffee_raw_volume/users',
    format => 'csv',
    header => true
  )
);
```

```sql
-- distinct user_id comparison
SELECT
  'bronze_table' AS source_type,
  COUNT(DISTINCT user_id) AS distinct_user_ids
FROM coffee.bronze.users

UNION ALL

SELECT
  'raw_csv_volume' AS source_type,
  COUNT(DISTINCT user_id) AS distinct_user_ids
FROM (
  SELECT *
  FROM read_files(
    '/Volumes/workspace/default/coffee_raw_volume/users',
    format => 'csv',
    header => true
  )
);
```

```sql
-- null check on user id in bronze
SELECT
  COUNT(*) AS null_user_id_count
FROM coffee.bronze.users
WHERE user_id IS NULL;
```

```sql
-- confirming that data is uploaded from 24 different files
SELECT
  COUNT(DISTINCT source_file) AS distinct_source_files
FROM coffee.bronze.users;
```
