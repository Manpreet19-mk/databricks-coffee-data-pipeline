####Unit Testing – Bronze vouchers
This notebook validates the Bronze ingestion of vouchers data by comparing the ingested Bronze table with the raw source files stored in Databricks Volumes.

The tests focus on:

- Row count consistency
- Null checks
- Duplicate detection

```sql

-- Row count comparison between Bronze table and original CSV source
SELECT
  'bronze_table' AS source_type,
  COUNT(*) AS row_count
FROM coffee.bronze.vouchers

UNION ALL

SELECT
  'raw_csv_volume' AS source_type,
  COUNT(*) AS row_count
FROM (
  SELECT *
  FROM read_files(
    '/Volumes/workspace/default/coffee_raw_volume/vouchers',
    format => 'csv',
    header => true
  )
);
```

```sql
--null check
SELECT
  COUNT(*) AS null_voucher_code_count
FROM coffee.bronze.vouchers
WHERE voucher_code IS NULL;
```

```sql
--duplicate check
SELECT
  voucher_code,
  valid_from,
  valid_to,
  COUNT(*) AS cnt
FROM coffee.bronze.vouchers
GROUP BY voucher_code, valid_from, valid_to
HAVING COUNT(*) > 1;
```
